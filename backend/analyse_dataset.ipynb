```python
import pandas as pd
import networkx as nx
import json
import igraph as ig
import leidenalg
```

```python
with open("data/MC3_graph.json") as f:
    data = json.load(f)

nodes = data["nodes"]
edges = data["edges"]

```

```python
print(f"Length Nodes: {len(nodes)}, Length Edges: {len(edges)}")
print(nodes[0])
print(edges[0])
```

```python
G = nx.Graph()

for node in nodes:
    node_id = node["id"]
    attrs = {k: v for k, v in node.items() if k != "id"}
    G.add_node(node_id, **attrs)

for edge in edges:
    source = edge["source"]
    target = edge["target"]
    attrs = {k: v for k, v in edge.items() if k not in ["source", "target"]}
    G.add_edge(source, target, **attrs)
```

```python

nodes = list(G.nodes())
edges = list(G.edges())
g = ig.Graph()
g.add_vertices(nodes)
g.add_edges(edges)

# Leiden ausführen
partition = leidenalg.find_partition(g, leidenalg.ModularityVertexPartition)
```

```python
import matplotlib.pyplot as plt

community_map = {}
for comm_id, community in enumerate(partition):
    for node in community:
        community_map[g.vs[node]["name"]] = comm_id

node_colors = [community_map[node] for node in G.nodes()]
node_sizes = [G.degree[node] * 30 for node in G.nodes()]  

# 5. Zeichnen
plt.figure(figsize=(10, 8))
pos = nx.forceatlas2_layout(G, seed=42)
nx.draw(
    G, pos,
    node_color=node_colors,
    node_size=node_sizes,
    cmap=plt.cm.tab10
)
plt.title("Leiden Community Detection")
plt.show()
```
